```python
# 끝말 잇기 게임

print("끝말잇기 게임")
print("시작")

var = "가나다"
print(var)

user_word = [var]

while True: 
    user_input = input("단어를 입력해주세요 : ")

    if user_input[0] == var[-1] and user_input not in user_word:
        var = user_input
        user_word.append(user_input)
    else:
        print("땡")
        break
```

```python
# 끝말 잇기 게임

print("끝말잇기 게임")
print("시작")

var = "가나다"
var_list = [var] ###
print(var)

while True: 
    answer = input("끝말잇기 단어 입력: ")
    print(answer) 
    if answer == "":
        print('똑바로 입력하고 있는데~~')
        continue
    if answer in var_list: ###
        print('떙')
        break
    if var[-1] == answer[0]:
        var_list.append(answer) ###
        var = answer # 업데이트
    else:
        print('땡')
        break

print('게임 종료')
```

```python
addr = {
    '박지민': '010-1111-1234', '전정국': '010-2222-1234',
    '민윤기': '010-3333-1234', '정호석': '010-4444-1234',
    '김태형': '010-5555-1234', '김석진': '010-6666-1234'
}

while True:
    menu = input("메뉴 선택(1: 추가, 2: 검색, 0: 종료)? ")

    if menu == '1':
        name = input("친구 이름 입력: ")
        phone = input("존화번호 입력: ")
        addr[name] = phone

    elif menu == '2':
        print(f"검색 가능한 친구들: {list(addr.keys())}")
        search_name = input("친구 이름 입력: ")

        if search_name in addr:
            print(f"{search_name}의 연락처: {addr[search_name]}")
        else:
            print(f"{search_name}은(는) 연락처에 없습니다.")

    elif menu == '0':
        print("프로그램을 종료합니다!")
        break

    else:
        print("메뉴 선택 오류입니다. 다시 선택하세요.")
```
