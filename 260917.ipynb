## 클래스

```python
class Student:
    population = 0 # 클래스 변수 (공유)

    def __init__(self, name, age): # 생성자
        self.name = name # 인스턴스 변수
        self.age = age
        Student.population += 1

    # 매서드
    def hello(self):
        print(f'{self.name}님 안녕하세요.')
```

```python
# 상속
class NewStudent(Student):
    population = 0 # 클래스 변수 (공유)

    # 새로운 메서드
    def study(self):
        print(f'{self.name}님 공부하세요.')

    # 메서드 오버라이딩 (메서드 재정의)
    def hello(self):
        print(f'{self.name}님 신~나게 안녕하세요.')
```

```python
std3 = NewStudent('정윤교', 25)
```

```python
std3.hello()
```

```python
std1 = Student('임원묵', 26)
```

```python
std1.population
```

```python
std2 = Student('최동건', 25)
```

```python
std1.population
```

```python
std2.hello() # name님 안녕하세요
```
