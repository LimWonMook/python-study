```python
marks = [90, 25, 67, 45, 80]

number = 0

for mark in marks:
    number += 1

    if mark >= 60:
        print(f"{number}번 학생은 합격입니다.")
    else:
        print(f"{number}번 학생은 불합격입니다.")
```

```python
# 끝말 잇기 게임

print("끝말잇기 게임")
print("시작")

var = "가나다"
print(var)

user_word = [var]

while True: 
    user_input = input("단어를 입력해주세요 : ")

    if user_input[0] == var[-1] and user_input not in user_word:
        var = user_input
        user_word.append(user_input)
    else:
        print("땡")
        break
```

```python
# 끝말 잇기 게임

print("끝말잇기 게임")
print("시작")

var = "가나다"
var_list = [var] ###
print(var)

while True: 
    answer = input("끝말잇기 단어 입력: ")
    print(answer) 
    if answer == "":
        print('똑바로 입력하고 있는데~~')
        continue
    if answer in var_list: ###
        print('떙')
        break
    if var[-1] == answer[0]:
        var_list.append(answer) ###
        var = answer # 업데이트
    else:
        print('땡')
        break

print('게임 종료')
```
