```python
# 성적 처리 프로그램
# 90점 이상 A, 80점 이상 B, 70점 이상 C, 60점 이상 D, 60점 미만 F
```

```python
score = int(input("성적를 입력하세요: "))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

```python
input_num = 0

while True:
    print(input_num)
    input_num = int(input("원하는 메뉴의 번호를 입력하세요: "))
    if input_num == 4:
        print("프로그램을 종료합니다.")
        break
```

```python
# 20보다 작은 정수 중에서 3의 배수와 5의 배수를 제외하고 출력

x=0

while True:
    x = x + 1
    if x > 20:
        break
    if x % 3 == 0 or x % 5 == 0:
        continue

    print(x, end = " ")
```

```python
# 미션1. 숫자 맞추기
# '너무 작아요', '너무 커요', '정답입니다'를 출력 / 정답 맞추면 게임 종료
# 몇번만에 맞췄는지 출력 / 10번 안에 못맞추면 강제 게임 종료 기능 추가

import random 

number = random.randint(1, 100)
count = 0

while True:
    guess = int(input("1부터 100 사이의 숫자를 입력하세요: "))
    count += 1
    
    if guess < number:
        print('너무 작아요')
    elif guess > number:
        print('너무 커요')
    else:
        print(f'정답입니다! {count}번 만에 맞췄습니다.')

    if count >= 10:
        print(f'게임 종료! 정답은 {number}였습니다.')
        break
```
